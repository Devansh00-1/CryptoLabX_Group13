# Part A — MT19937 Stream Cipher

This lab uses Python's built-in `random.Random` MT19937 generator to generate one byte per plaintext byte. XOR makes the same routine work for encryption and decryption. **This construction is educational and is not secure.**

MT19937 is a deterministic pseudorandom generator with a 19937-bit state. Python's `random` module is intended for simulation and does not provide cryptographic security. Generator output can be predictable; it should not be used as a cryptographic keystream.

A known-plaintext attack uses a known plaintext segment to test key or state candidates. The target message has a random printable prefix of 5–20 characters followed by exactly fourteen `A` characters. The attacker knows the marker but not its location, so each full candidate decryption is searched for `A` repeated 14 times. The 16-bit seed space has 2^16 = 65,536 candidates, making exhaustive search practical.


## Recovery pseudocode

```text
for seed from 0 through 65535:
    candidate_plaintext = XOR(ciphertext, MT19937_bytes(seed, ciphertext_length))
    if candidate_plaintext contains fourteen consecutive 'A' bytes:
        return seed, candidate_plaintext
fail: no matching seed
```

The implementation returns the marker offset for reporting only; it is not supplied to the search.
